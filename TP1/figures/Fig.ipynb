```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rc

rc('font', size = 18)
```

```python
h = lambda x : (1.0584 + 0.3181 * x)/(0.6794 + 0.01776 * x)

pH = np.linspace(0,14, num = 50)

h(6.8)
```

```python
fig = plt.figure(figsize = (12,8))
ax = fig.subplots()
ax.plot(pH, h(pH), color = 'black')
ax.grid()

ax.set_ylabel("Taux d'hydrolyse $k$ [$\emptyset$]")
ax.set_xlabel("$pH$")
#ax.axhline(y = 2.5, color = 'red')
#ax.axhline(y = 3.5, color = 'red')
#ax.axhline(y = 4.5, color = 'red')
#ax.axhline(y = 5.5, color = 'red')

fig.savefig("Corr03.png")
```

```python
h = lambda x, pKe = 14 : (3.2507 - 0.1566 * pKe +  0.3131 * pH)/(0.80168 - 0.00874 * pKe +  0.01749 * pH)

pH = np.linspace(0,14, num = 50)
```

```python
colors = ("gray", 'darkgray', 'gainsboro')
pKe = [12,13,14]

pKe_colors = dict(zip(pKe, colors))

fig = plt.figure(figsize = (12,8))
ax = fig.subplots()
for pKe_n in pKe :
    ax.plot(pH, h(pH, pKe = pKe_n), color = pKe_colors[pKe_n], label = 'pKe = {}'.format(pKe_n))

ax.grid()
ax.axhline(y=4, color = 'red')

ax.set_xlabel("$pH$")
ax.set_ylabel("$h$")

ax.legend()

fig.savefig("Corr05.png")
```

```python
DrH = 13430 * 4.184 #± 5 cal mol–1
R = 8.314

K = lambda T, Tamb = 298, Ke = 10**(-14) : -np.log10(Ke * np.exp(DrH / R * (1 / Tamb - 1 / T)))

T = 273 + np.linspace(25, 200, num = 50)

fig = plt.figure(figsize = (12,8))
ax = fig.subplots()
ax.plot(T, K(T), color = 'black')

ax.grid()

ax.set_xlabel("$T$ [$K$]")
ax.set_ylabel("$pK_e (T)$ [$\emptyset$]")

fig.savefig("Corr04.png")
```
